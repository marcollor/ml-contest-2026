```python
from pathlib import Path
from subprocess import run
from sys import executable
from tempfile import TemporaryDirectory

materials = Path('/content/ml-contest-materials')
with TemporaryDirectory(prefix='temporalcnn-check-') as directory:
    submission = Path(directory) / 'submission'
    run(['git', 'clone', 'https://github.com/marcollor/ml-contest-2026.git', submission], check=True)
    run(['git', 'log', '-1', '--format=Evaluating commit %H'], cwd=submission, check=True)
    run([executable, '-m', 'pip', 'install', '-r', submission / 'requirements.txt'], check=True)
    run([executable, 'harness.py', 'evaluate', '--model', submission / 'model.py'], cwd=materials, check=True)
```
